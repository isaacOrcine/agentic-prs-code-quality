# 02 · Clonagem dos repositórios do piloto

**Objetivo.** Baixar o histórico (apenas o branch padrão, sem tags) dos cinco projetos do piloto para as etapas de atribuição e medição.

- **Lê:** `../out/piloto.csv`
- **Grava:** um clone por projeto em `../repos/<dono>__<repositório>/` (via `git clone` do GitHub, o que exige rede)

**O que esta célula faz.** Para cada projeto de `piloto.csv`, clona o repositório do GitHub em `../repos/` (só o branch padrão, sem tags), pulando os que já existem. O histórico clonado é a base de todas as etapas seguintes.

```python
import pandas as pd, subprocess, os
for nome in pd.read_csv('../out/piloto.csv').full_name:
    destino = '../repos/' + nome.replace('/', '__')
    if not os.path.exists(destino):
        subprocess.run(['git', 'clone', '--single-branch', '--no-tags',
                        f'https://github.com/{nome}', destino], check=True)
```

**O que isto mostra.** Foram iniciados cinco clones, um por projeto do piloto: numbagg, HARK, DDNS, diffyscan e PyAirbyte. Como a chamada usa `check=True`, uma falha interromperia a célula, então a ausência de erro na saída indica que os cinco clones terminaram. A saída não informa número de commits nem tamanho dos repositórios.
